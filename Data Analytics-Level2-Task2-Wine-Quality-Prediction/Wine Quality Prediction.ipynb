```python
# Features
X = df.drop(["quality", "quality_category"], axis=1)

# Target
y = df["quality_category"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling completed.")
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)

print("Random Forest predictions completed.")
```

```python
from sklearn.metrics import accuracy_score
```

```python
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
from sklearn.metrics import classification_report

print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=rf_model.classes_,
    yticklabels=rf_model.classes_
)

plt.title("Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
sgd_model = SGDClassifier(
    random_state=42,
    max_iter=1000,
    tol=1e-3
)

sgd_model.fit(X_train_scaled, y_train)
```

```python
y_pred_sgd = sgd_model.predict(X_test_scaled)

print("SGD predictions completed.")
```

```python
print("SGD Classification Report:")
print(classification_report(y_test, y_pred_sgd))
```

```python
cm_sgd = confusion_matrix(y_test, y_pred_sgd)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_sgd,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=sgd_model.classes_,
    yticklabels=sgd_model.classes_
)

plt.title("SGD Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from sklearn.svm import SVC
```

```python
svc_model = SVC(
    random_state=42
)

svc_model.fit(X_train_scaled, y_train)
```

```python
y_pred_svc = svc_model.predict(X_test_scaled)

print("SVC predictions completed.")
```

```python
svc_accuracy = accuracy_score(y_test, y_pred_svc)

print("SVC Accuracy:", svc_accuracy)
```

```python
print("SVC Classification Report:")
print(classification_report(y_test, y_pred_svc))
```

```python
cm_svc = confusion_matrix(y_test, y_pred_svc)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_svc,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=svc_model.classes_,
    yticklabels=svc_model.classes_
)

plt.title("SVC Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python


feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)

plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Chemical Feature")

plt.show()
```

```python
# Final comparison table

comparison = pd.DataFrame({
    "Model": ["Random Forest", "SGD Classifier", "SVC"],
    
    "Accuracy": [
        rf_accuracy,
        sgd_accuracy,
        svc_accuracy
    ],
    
    "Precision": [
        rf_report["weighted avg"]["precision"],
        sgd_report["weighted avg"]["precision"],
        svc_report["weighted avg"]["precision"]
    ],
    
    "Recall": [
        rf_report["weighted avg"]["recall"],
        sgd_report["weighted avg"]["recall"],
        svc_report["weighted avg"]["recall"]
    ],
    
    "F1-Score": [
        rf_report["weighted avg"]["f1-score"],
        sgd_report["weighted avg"]["f1-score"],
        svc_report["weighted avg"]["f1-score"]
    ]
})

comparison.round(3)
```

```python
comparison = comparison.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

comparison.round(3)
```

```python
## Conclusion

In this project, three classification models — Random Forest, SGD Classifier, and Support Vector Classifier (SVC) — were trained to predict wine quality categories (Low, Medium, and High).

Among the three models, Random Forest achieved the best overall performance with an accuracy of 86.0% and a weighted F1-score of 0.841. SVC achieved an accuracy of 84.3%, while SGD Classifier achieved 82.1%.

Random Forest is therefore the most suitable model for deployment for this dataset because it provided the highest accuracy, precision, recall, and F1-score. It also provides feature importance values, which makes it useful for understanding which chemical properties contribute most to wine quality prediction.

Although the model performs well, the dataset contains class imbalance, particularly fewer Low and High quality wines compared with Medium quality wines. Therefore, additional techniques such as class balancing or collecting more samples from underrepresented categories could be considered before production deployment.

### Business Recommendations

1. Use Random Forest as the preferred model for predicting wine quality categories.
2. Focus on the chemical features identified as most important by the Random Forest model when analysing wine quality.
3. Collect more samples for underrepresented quality categories to improve prediction performance.
4. Monitor model performance regularly when new wine-quality data becomes available.
```
