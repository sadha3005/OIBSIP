```python
print("Mean:", df['Sales'].mean())
print("Median:", df['Sales'].median())
print("Mode:", df['Sales'].mode()[0])
print("Standard Deviation:", df['Sales'].std())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
```

```python
print(df['Order Date'].dtype)
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python
df[['Order Date', 'Month']].head()
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
```

```python
monthly_sales
```

```python
plt.figure(figsize=(12, 6))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
df['Quarter'] = df['Order Date'].dt.to_period('Q')
```

```python
df[['Order Date', 'Quarter']].head()
```

```python
quarterly_sales = df.groupby('Quarter')['Sales'].sum()
```

```python
quarterly_sales
```

```python
plt.figure(figsize=(10, 5))

plt.plot(quarterly_sales.index.astype(str), quarterly_sales.values, marker='o')

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
### Observation – Quarterly Sales Trend

* Sales fluctuated across the quarters from 2015 to 2018.
* The highest quarterly sales were recorded in **2017 Q1**, with approximately **95,293** in sales.
* **2018 Q1** also recorded strong sales of approximately **90,536**.
* The lowest quarterly sales were recorded in **2016 Q4**, at approximately **34,193**.
* Overall, the quarterly trend shows that sales increased significantly in 2017 and remained relatively strong in 2018.
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

segment_sales
```

```python
plt.figure(figsize=(8, 5))

plt.bar(segment_sales.index, segment_sales.values)

plt.title('Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python
### Observation – Sales by Customer Segment

* The **Consumer** segment generated the highest sales, approximately **1.15 million**.
* The **Corporate** segment generated approximately **0.69 million** in sales.
* The **Home Office** segment generated approximately **0.42 million** in sales.
* This indicates that the Consumer segment is the largest contributor to overall sales.
```

```python
top_10_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_10_products
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top_10_products.index, top_10_products.values)

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')

plt.gca().invert_yaxis()
plt.tight_layout()

plt.show()
```

```python
### Observation – Top 10 Products by Sales

* The **Canon imageCLASS 2200 Advanced Copier** generated the highest sales among the top 10 products, with approximately **61,600** in sales.
* The **Fellowes PB500 Electric Punch Plastic Comb Binding Machine** ranked second, with approximately **27,453**.
* The **Cisco TelePresence System EX90 Videoconferencing Unit** ranked third, with approximately **22,638**.
* The top-selling products are mainly **office equipment and technology-related products**.
* The Canon copier has significantly higher sales than the other top 10 products.
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales
```

```python
plt.figure(figsize=(8, 5))

plt.bar(category_sales.index, category_sales.values)

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')

plt.tight_layout()
plt.show()
```

```python
### Observation – Revenue by Category

* **Technology** generated the highest revenue, approximately **827,456**.
* **Furniture** generated approximately **728,659** in revenue.
* **Office Supplies** generated approximately **705,422** in revenue.
* Technology is the strongest-performing category and contributes the largest share of total sales.
* The difference between the three categories is relatively moderate, showing that all categories contribute significantly to overall revenue.
```

```python
correlation = df.select_dtypes(include='number').corr()

plt.figure(figsize=(8, 6))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap')

plt.tight_layout()
plt.show()
```

```python
### Observation – Correlation Heatmap

* The heatmap shows the correlation between the numerical variables in the dataset.
* **Sales** has very weak correlation with **Row ID** and **Postal Code**.
* This indicates that these numerical identifier/location fields do not have a meaningful linear relationship with sales.
* The heatmap also shows that most numerical variables have weak relationships with each other.
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df['Sales'], bins=30)

plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Number of Orders')

plt.tight_layout()
plt.show()
```

```python
### Observation – Sales Distribution

* Most orders have relatively low sales values.
* The distribution is **right-skewed**, with a small number of orders having very high sales values.
* These high-value orders increase the average sales compared with the median sales.
* This indicates that a few large transactions contribute significantly to total revenue.
```

```python
## Conclusion

The exploratory data analysis provides useful insights into the retail sales performance. Sales show fluctuations across months and quarters, with particularly strong performance in 2017 and 2018. The Consumer segment contributes the highest sales, while Technology is the highest-revenue product category. The Top 10 Products analysis also shows that a small number of products generate substantially higher sales than others. The sales distribution is right-skewed, indicating that a few high-value transactions have a significant impact on overall revenue.

## Business Recommendations

1. **Focus on high-performing Technology products:** Increase inventory availability and promotional activities for top-performing Technology products to capture more sales opportunities.

2. **Strengthen the Consumer segment:** Since Consumers generate the highest sales, targeted offers, loyalty programs, and personalized marketing campaigns can help increase repeat purchases.

3. **Promote high-value products and transactions:** Identify products associated with high-value orders and use bundles, cross-selling, and targeted promotions to increase average order value.
```
